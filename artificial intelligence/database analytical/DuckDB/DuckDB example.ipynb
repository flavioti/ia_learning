```python
import duckdb
import pandas as pd

# Create a sample Pandas DataFrame
data = {
    'item': ['Apple', 'Banana', 'Orange', 'Apple', 'Banana'],
    'category': ['Fruit', 'Fruit', 'Fruit', 'Fruit', 'Fruit'],
    'price': [1.2, 0.5, 0.8, 1.3, 0.6],
    'quantity': [10, 15, 7, 20, 12]
}
df = pd.DataFrame(data)

# Use DuckDB to run a SQL query directly on the Pandas DataFrame variable 'df'
result = duckdb.sql("""
    SELECT item, SUM(price * quantity) AS total_sales
    FROM df
    GROUP BY item
    ORDER BY total_sales DESC
""").df() # Convert the DuckDB result back to a Pandas DataFrame

print(result)
```
